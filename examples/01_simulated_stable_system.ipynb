# Stable known-system perturbation
Synthetic examples are not empirical brain findings.

```python
import numpy as np
from nssl.dynamics import LinearContinuousDynamics
from nssl.simulation import system_matrix
from nssl.metrics import local_trajectory_separation, recovery_fraction
```

```python
t=np.linspace(0,5,201)
model=LinearContinuousDynamics(system_matrix("stable"))
initial=np.array([[1.,.5]])
result=model.compare(initial, initial+np.array([[.4,0.]]),t)
d=local_trajectory_separation(result.perturbed,result.baseline)
print(d[0,0],d[0,-1],recovery_fraction(d)[0])
```

```python
import matplotlib.pyplot as plt
fig, ax=plt.subplots()
ax.plot(t,d[0],label="Exact synthetic trajectory separation")
ax.set(xlabel="Time (a.u.)",ylabel="Euclidean distance",title="Synthetic stable system")
ax.legend()
plt.show()
```

# Regression-to-the-mean trap
Synthetic examples are not empirical brain findings.

```python
import numpy as np
import matplotlib.pyplot as plt
from nssl.simulation import regression_to_mean_trap
```

```python
result=regression_to_mean_trap(n_trials=1000,random_state=42)
print("Synthetic apparent recovery:",result["mean_apparent_recovery"])
```

```python
x=result["series"][result["selected"]]
k=result["index"]
plt.plot(range(-5,12),np.abs(x[:,k-5:k+12]).mean(axis=0))
plt.axvline(0,linestyle="--")
plt.xlabel("Sample relative to extreme");plt.ylabel("Mean absolute deviation")
plt.title("Synthetic stationary AR(1) selection artifact")
plt.show()
```
